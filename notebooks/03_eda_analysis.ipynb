## Univariate Analysis

```python
plt.figure(figsize=(8,5))
sns.histplot(df['age'], bins=20, kde=True)

plt.title('Customer Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')

plt.show()
```

```python
df['age'].value_counts()
```

```python
ax = sns.countplot(x='gender', data=df)

ax.bar_label(ax.containers[0])

plt.title("Gender Distribution")
plt.show()
```

```python
sns.countplot(y='category', data=df)
plt.title("Product Categories")
plt.show()
```

```python
sns.boxplot(x=df['purchase_amount_usd'])
plt.title("Purchase Amount")
plt.show()
```

## Bivariate Analysis

Gender vs Purchase Amount

```python
sns.boxplot(x='gender', y='purchase_amount_usd', data=df)
plt.title("Gender vs Purchase Amount")
plt.show()
```

Category vs Review Rating

```python
plt.figure(figsize=(10,5))
sns.barplot(x='category', y='review_rating', data=df)
plt.xticks(rotation=45)
plt.title("Category vs Review Rating")
plt.show()
```

Season vs Sales

```python
sns.countplot(x='season', data=df)
plt.title("Seasonal Purchases")
plt.show()
```

Loyalty Segment

```python
sns.countplot(x='loyalty_segment', data=df)
plt.title("Customer Loyalty Segments")
plt.show()
```

Retention Risk

```python
ax = sns.countplot(x='retention_risk', data=df)

ax.bar_label(ax.containers[0])

plt.title("Retention Risk")
plt.show()
```

Premium Customers

```python
ax = sns.countplot(x='premium_customer', data=df)

ax.bar_label(ax.containers[0])

plt.title("Premium Customers")
plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.barplot(
    x='discount_applied',
    y='purchase_amount_usd',
    data=df,
    estimator='mean'
)

plt.title("Average Purchase Amount by Discount")
plt.xlabel("Discount Applied")
plt.ylabel("Average Purchase Amount (USD)")

plt.show()
```

```python
plt.figure(figsize=(8,5))

ax = sns.countplot(x='payment_method', data=df)

ax.bar_label(ax.containers[0])

plt.title("Payment Method Distribution")
plt.xlabel("Payment Method")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)

plt.show()
```
